```python
# ACTIVIDAD 2-a: Ingeniería de características (RFM)

print("\n CÁLCULO RFM (Actividad 2-a) ")

# 1 y 2. Definir fecha snapshot (un día después de la última transacción)
snapshot_date = df_merged['transaction_date'].max() + pd.Timedelta(days=1)

# Agrupar por cliente para calcular las 3 métricas RFM
rfm = df_merged.groupby('customer_id').agg({
    'transaction_date': lambda x: (snapshot_date - x.max()).days, # Recency: Días desde la última compra
    'transaction_id': 'nunique',                                  # Frequency: Número de tickets/compras
    'line_item_amount': 'sum'                                     # Monetary: Dinero total gastado
}).reset_index()

# Renombrar las columnas para que quede estándar
rfm.rename(columns={
    'transaction_date': 'Recency',
    'transaction_id': 'Frequency',
    'line_item_amount': 'Monetary'
}, inplace=True)

#  3.Eliminar outliers de 'Monetary' basándose en el percentil 95
percentil_95 = rfm['Monetary'].quantile(0.95)
rfm_clean = rfm[rfm['Monetary'] <= percentil_95].copy()

print(f"2-a (Paso 3): Límite del percentil 95 para Monetary: {percentil_95:.2f}")
print(f"2-a (Paso 3): Clientes totales originales: {len(rfm)}")
print(f"2-a (Paso 3): Clientes tras eliminar outliers: {len(rfm_clean)}\n")

# 4. Visualizar las distribuciones de RFM con histogramas interactivos
fig_r = px.histogram(rfm_clean, x='Recency', title='2-a: Distribución de Recency (Días desde la última compra)', nbins=30)
fig_f = px.histogram(rfm_clean, x='Frequency', title='2-a: Distribución de Frequency (Número de tickets)', nbins=30)
fig_m = px.histogram(rfm_clean, x='Monetary', title='2-a: Distribución de Monetary (Gasto total)', nbins=30)

fig_r.show()
fig_f.show()
fig_m.show()
```

```python
from sklearn.cluster import KMeans
import warnings
warnings.filterwarnings('ignore') # Ocultamos advertencias de KMeans

print(" 2-a: CLUSTERING RFM ")

# 5. Elbow Method (Visualización)
# (Solo como buena práctica, calculamos la inercia para 'Monetary' como ejemplo)
inercia = []
for k in range(1, 10):
    kmeans = KMeans(n_clusters=k, random_state=42, n_init=10)
    kmeans.fit(rfm_clean[['Monetary']])
    inercia.append(kmeans.inertia_)

fig_elbow = px.line(x=range(1, 10), y=inercia, title='2-a (Paso 5): Elbow Method para Monetary',
                    labels={'x': 'Número de Clústeres (k)', 'y': 'Inercia (WCSS)'})
fig_elbow.show()

# 6. Clustering con KMeans (k=3) y mapeo de etiquetas (1 a 3)
# Función para ordenar los clusters.
# Para Frequency y Monetary, mayor valor = mejor score (3).
# Para Recency, menor valor (más reciente) = mejor score (3).
def order_cluster(cluster_field_name, target_field_name, df, ascending):
    new_cluster_field_name = 'new_' + cluster_field_name
    df_new = df.groupby(cluster_field_name)[target_field_name].mean().reset_index()
    df_new = df_new.sort_values(by=target_field_name, ascending=ascending).reset_index(drop=True)
    df_new['index'] = df_new.index
    df_final = pd.merge(df, df_new[[cluster_field_name, 'index']], on=cluster_field_name)
    df_final = df_final.drop([cluster_field_name], axis=1)
    df_final = df_final.rename(columns={"index": cluster_field_name})
    # Sumamos 1 para que los scores vayan de 1 a 3 (en lugar de 0 a 2)
    df_final[cluster_field_name] = df_final[cluster_field_name] + 1
    return df_final

# Aplicamos KMeans a cada variable
kmeans = KMeans(n_clusters=3, random_state=42, n_init=10)

# Recency (El orden es inverso: más reciente es mejor, por lo que ascending=False para asignar 3 a los de menor recencia)
rfm_clean['R_Cluster'] = kmeans.fit_predict(rfm_clean[['Recency']])
rfm_clean = order_cluster('R_Cluster', 'Recency', rfm_clean, False)

# Frequency (Mayor frecuencia es mejor)
rfm_clean['F_Cluster'] = kmeans.fit_predict(rfm_clean[['Frequency']])
rfm_clean = order_cluster('F_Cluster', 'Frequency', rfm_clean, True)

# Monetary (Mayor monetario es mejor)
rfm_clean['M_Cluster'] = kmeans.fit_predict(rfm_clean[['Monetary']])
rfm_clean = order_cluster('M_Cluster', 'Monetary', rfm_clean, True)


# 7. TotalScore y Segmentación ('Low', 'Medium', 'High')

rfm_clean['TotalScore'] = rfm_clean['R_Cluster'] + rfm_clean['F_Cluster'] + rfm_clean['M_Cluster']

# Segmentamos basados en el TotalScore (que va de 3 a 9)
def segment_customer(score):
    if score >= 8:
        return 'High'
    elif score >= 5:
        return 'Medium'
    else:
        return 'Low'

rfm_clean['Segment'] = rfm_clean['TotalScore'].apply(segment_customer)



# 8. Mostrar filas y estadísticas medias por segmento
print("\n 2-a (Paso 8): Primeras filas del DataFrame RFM con Scores ")
display(rfm_clean.head())

print("\n 2-a (Paso 8): Estadísticas medias por segmento ")
segment_stats = rfm_clean.groupby('Segment').agg({
    'Recency': 'mean',
    'Frequency': 'mean',
    'Monetary': ['mean', 'count']
}).round(2)
display(segment_stats)

# Visualización extra para ver el tamaño de los segmentos
fig_segments = px.pie(rfm_clean, names='Segment', title='Distribución de Clientes por Segmento')
fig_segments.show()
```

Estrategia para el segmento "High" (Fidelización y Recompensa):
Dado que vienen cada menos de 3 días y gastan mucho, el objetivo no es convencerlos de que compren, sino recompensar su lealtad. Se podría proponer un programa de puntos VIP ("café gratis a la décima compra") o invitarles a probar nuevos productos antes que nadie.

Estrategia para el segmento "Medium" (Up-selling y Venta Cruzada):
Como son el grupo más grande, cualquier pequeña mejora aquí disparará los ingresos. La estrategia debe centrarse en aumentar el ticket medio (Monetary). Se pueden ofrecer combos  o promociones por tiempo limitado para empujarlos al segmento "High".

Estrategia para el segmento "Low" (Reactivación):
Estos clientes están a punto de perderse (llevan 14 días sin ir). Se requiere una estrategia agresiva de reactivación mediante campañas de email/SMS con fuertes descuentos (ej. "¡Te echamos de menos! Vuelve esta semana y tienes un 30% en tu compra") o encuestas para saber por qué dejaron de ir.

```python
import xgboost as xgb
import lightgbm as lgb
from sklearn.ensemble import RandomForestClassifier
from sklearn.model_selection import train_test_split
from sklearn.metrics import classification_report, confusion_matrix, roc_auc_score, roc_curve
import pandas as pd

print(" 2-b: PREPARANDO DATOS PARA CLASIFICACIÓN ")

# 1. Fechas límite del dataset
fecha_min = df_merged['transaction_date'].min()
fecha_max = df_merged['transaction_date'].max()
dias_totales = (fecha_max - fecha_min).days

print(f"Fecha inicio: {fecha_min.date()}, Fecha fin: {fecha_max.date()}")
print(f"Total días: {dias_totales}")

# 2. Definir período de calibración (80%) y ventana hold-out (20%)
dias_calibracion = int(dias_totales * 0.8)
fecha_corte = fecha_min + pd.Timedelta(days=dias_calibracion)

print(f"Fecha de corte (80%): {fecha_corte.date()}")

# Separamos las transacciones temporalmente
df_calibracion = df_merged[df_merged['transaction_date'] <= fecha_corte]
df_holdout = df_merged[df_merged['transaction_date'] > fecha_corte]

# Calculamos RFM estrictamente para el período de calibración
rfm_calib = df_calibracion.groupby('customer_id').agg({
    'transaction_date': lambda x: (fecha_corte - x.max()).days,
    'transaction_id': 'nunique',
    'line_item_amount': 'sum'
}).reset_index()

rfm_calib.rename(columns={
    'transaction_date': 'Recency',
    'transaction_id': 'Frequency',
    'line_item_amount': 'Monetary'
}, inplace=True)

# 3. Crear variable objetivo (Target)
# 1 = Compró en el hold-out, 0 = No compró
clientes_holdout = df_holdout['customer_id'].unique()
rfm_calib['compro_holdout'] = rfm_calib['customer_id'].apply(lambda x: 1 if x in clientes_holdout else 0)

print("\nDistribución de la variable objetivo (1=Compró, 0=No compró):")
display(rfm_calib['compro_holdout'].value_counts(normalize=True) * 100)

# Preparar X (features) e y (target)
X = rfm_calib[['Recency', 'Frequency', 'Monetary']]
y = rfm_calib['compro_holdout']

# Dividir en Train, Validation y Test (30% Test; del resto, 20% Validación)
X_temp, X_test, y_temp, y_test = train_test_split(X, y, test_size=0.30, random_state=42)
X_train, X_val, y_train, y_val = train_test_split(X_temp, y_temp, test_size=0.20, random_state=42)

print("\nTamaños de los conjuntos:")
print(f"Train: {X_train.shape[0]} | Validation: {X_val.shape[0]} | Test: {X_test.shape[0]}")
```

```python
# 2-b (Paso 4): Entrenar modelos y evaluar en validación
from sklearn.metrics import roc_curve, auc
import matplotlib.pyplot as plt
import warnings

# Silenciar warnings
warnings.filterwarnings("ignore")

print("Iniciando entrenamiento de modelos...\n")

# 1. Instanciar los modelos con una semilla fija (random_state) para reproducibilidad
rf_model = RandomForestClassifier(random_state=42)
xgb_model = xgb.XGBClassifier(random_state=42, eval_metric='logloss')
lgb_model = lgb.LGBMClassifier(random_state=42, verbose=-1)

# 2. Entrenar los modelos con el conjunto Train
rf_model.fit(X_train, y_train)
xgb_model.fit(X_train, y_train)
lgb_model.fit(X_train, y_train)

# 3. Predicciones sobre el conjunto de Validación (Clases y Probabilidades)
y_pred_rf = rf_model.predict(X_val)
y_pred_xgb = xgb_model.predict(X_val)
y_pred_lgb = lgb_model.predict(X_val)

y_prob_rf = rf_model.predict_proba(X_val)[:, 1]
y_prob_xgb = xgb_model.predict_proba(X_val)[:, 1]
y_prob_lgb = lgb_model.predict_proba(X_val)[:, 1]

# 4. Evaluación: Matrices de Confusión y Reportes de Clasificación
modelos = {
    'Random Forest': y_pred_rf,
    'XGBoost': y_pred_xgb,
    'LightGBM': y_pred_lgb
}

for nombre, y_pred in modelos.items():
    print(f"================ {nombre} ================")
    print("Matriz de Confusión:")
    print(confusion_matrix(y_val, y_pred))
    print("\nReporte de Clasificación:")
    print(classification_report(y_val, y_pred))

# 5. Curva ROC y cálculo del AUC (Área Bajo la Curva)
plt.figure(figsize=(10, 8))

# Calculamos y graficamos para cada modelo
fpr_rf, tpr_rf, _ = roc_curve(y_val, y_prob_rf)
plt.plot(fpr_rf, tpr_rf, label=f'Random Forest (AUC = {auc(fpr_rf, tpr_rf):.3f})')

fpr_xgb, tpr_xgb, _ = roc_curve(y_val, y_prob_xgb)
plt.plot(fpr_xgb, tpr_xgb, label=f'XGBoost (AUC = {auc(fpr_xgb, tpr_xgb):.3f})')

fpr_lgb, tpr_lgb, _ = roc_curve(y_val, y_prob_lgb)
plt.plot(fpr_lgb, tpr_lgb, label=f'LightGBM (AUC = {auc(fpr_lgb, tpr_lgb):.3f})')

# Línea base (modelo aleatorio)
plt.plot([0, 1], [0, 1], 'k--', label='Aleatorio (AUC = 0.500)')
plt.xlim([0.0, 1.0])
plt.ylim([0.0, 1.05])
plt.xlabel('Tasa de Falsos Positivos (FPR)')
plt.ylabel('Tasa de Verdaderos Positivos (TPR)')
plt.title('2-b: Curva ROC comparativa en Validación')
plt.legend(loc="lower right")
plt.grid(alpha=0.3)
plt.show()
```
